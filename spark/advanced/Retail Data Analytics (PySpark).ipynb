# Retail Data Wrangling and Analytics

This notebook focuses on providing analytics of a UK based, online retail store. The objective is to uncover meaningful insights that can support data-driven decision-making and improve business outcomes.

The data wrangling process includes handling missing values, parsing and formatting dates, and ensuring that all columns are in the appropriate data types for downstream analysis. From there, the analysis explores customer behavior over time, including patterns in monthly active users and overall sales performance. A key part of the project involves performing RFM (Recency, Frequency, Monetary) analysis to segment customers based on purchasing behavior. These insights provide a foundation for future analysis, including customer retention strategies and sales forecasting.

### Load Dataset and Import **Libraries**

```
# Import Modules
from pyspark.sql import functions as F

# Load DataFrame
retail_df = (
  spark.read.format('csv')
  .option('inferSchema', 'true')
  .option('header', 'true')
  .option('sep', ',')
  .load('/FileStore/tables/online_retail.csv')
)

display(retail_df)
```

### Clean and Format Data

```
from pyspark.sql.functions import count, col, date_format, round

# Remove entries will null values and verify count results
def check_null(df):
    df.agg( *[ count(c).alias(c) for c in df.columns ] ).show()

retail_df = retail_df.dropna(how='any')
check_null(retail_df)

# Add invoice total and month columns for analysis
retail_df = (
    retail_df.withColumn('InvoiceTotal', round(col('Price')*col('Quantity'), 2))
    .withColumn('Month', date_format(col('InvoiceDate'), 'yyyy-MM'))
)

display(retail_df)
```

### Total Invoice Amount Distribution

```
from pyspark.sql.functions import sum

# Calculate invoice totals
invoice_totals = (
    retail_df.groupBy('Invoice')
    .agg(sum('InvoiceTotal').alias('InvoiceTotal')) 
)

# Use 85th percentile to remove outliers
quantile_85 = (invoice_totals.approxQuantile('InvoiceTotal', [0.85], 0.01)[0])
invoice_totals = (invoice_totals.filter((col('InvoiceTotal') > 0) & (col('InvoiceTotal') < quantile_85)))

display(invoice_totals)
```

### Monthly Placed and Canceled Orders

```
# Count monthly placed orders
monthly_placed = (
  retail_df.dropDuplicates(['Invoice'])
  .filter(col('quantity') > 0)
  .groupBy('Month')
  .agg(count('*').alias('OrdersPlaced'))
)

# Count monthly cancelled orders
monthly_canceled = (
  retail_df.dropDuplicates(['Invoice'])
  .filter(col('quantity') < 0)
  .groupBy('Month')
  .agg(count('*').alias('OrdersCanceled'))
)

# Join data for visualization
monthly_pc = (
  monthly_placed.join(monthly_canceled, 'Month', 'inner')
  .orderBy('Month')
)

display(monthly_pc)
```

### Monthly Sales

```
# Calculate total sales by month
monthly_sales = (
    retail_df.groupBy('Month')
    .agg(round(sum('InvoiceTotal'), 2).alias('MonthlySales'))
    .orderBy('Month')
)

display(monthly_sales)
```

### Monthly Sales Growth

```
from pyspark.sql.functions import lag
from pyspark.sql.window import Window

# Calculate the change(%) in sales by month
window = Window.orderBy("Month")
monthly_sales_growth = (
    monthly_sales.withColumn("Growth (%)", ((col("MonthlySales") - lag("MonthlySales", 1).over(window)) / lag("MonthlySales", 1).over(window) * 100))
)

display(monthly_sales_growth)
```

### Monthly Active Users

```
# Count the number of unique users that placed orders each month
monthly_users = (
  retail_df.withColumn('Month', date_format(col('InvoiceDate'), 'yyyy-MM'))
  .dropDuplicates(['Month','Customer ID'])
  .groupBy('Month')
  .agg(count('Customer ID').alias('Active Users'))
  .orderBy('Month')
)

display(monthly_users)
```

### Monthly New/Existing Users

```
from pyspark.sql.functions import min, when

# Partition for finding customer's first order month
window = Window.partitionBy('Customer ID')

# Find new/existing user counts for each month
ne_users = (
  retail_df.withColumn('FirstMonth', min('Month').over(window))
  .withColumn('UserType',when(col('Month') == col('FirstMonth'), 'New').otherwise('Existing'))
  .dropDuplicates(['Customer ID', 'Month'])
  .groupBy('Month')
  .agg(
        count(when(col('UserType') == 'New', True)).alias('New Users'),
        count(when(col('UserType') == 'Existing', True)).alias('Existing Users'),
  )
  .orderBy('Month')
)

display(ne_users)
```

### RFM

```
from pyspark.sql.functions import max, current_timestamp, to_timestamp, countDistinct
from pyspark.sql.types import DoubleType

# Add Duration Column
today = date_format(current_timestamp(),('yyyy-MM-dd HH:mm'))
rfm = retail_df.withColumn('Duration', round((to_timestamp(today).cast('long') - col('InvoiceDate').cast('long'))/86400))

# Find Recency, Frequency, Monetary
recency = rfm.groupBy('Customer ID').agg(min('Duration').alias('Recency'))
frequency = rfm.groupBy('Customer ID').agg(countDistinct('Invoice').cast(DoubleType()).alias('Frequency'))
monetary = rfm.groupBy('Customer ID').agg(round(sum('InvoiceTotal'), 2).alias('Monetary'))

# Build RFM dataframe
rfm = (
    recency.join(frequency, 'Customer ID')
    .join(monetary, 'Customer ID')
)

display(rfm)
```

```
from pandas import to_numeric

# Statistical Summary
pandas_rfm = (rfm.toPandas()).apply(to_numeric)
pandas_rfm.describe().T
```

```
# RFM Segmentation - Segment customers into categories based on defined expectations of the business.

# Define Recency Score
def r_score(r_value):
    if r_value <= 5000:
        return 1
    elif r_value <= 5200:
        return 2
    elif r_value <= 5400:
        return 3
    else:
        return 4
    
# Define Frequency Score
def f_score(f_value):
    if f_value <= 1: 
        return 1
    elif f_value <= 4:
        return 2
    elif f_value <= 7:
        return 3
    else:
        return 4
    
# Define Monetary Score
def m_score(m_value):
    if m_value <= 100:
        return 4
    elif m_value <= 1000:
        return 3
    elif m_value <= 2000:
        return 2
    else:
        return 1


from pyspark.sql.functions import udf
from pyspark.sql.types import StringType

r_udf = udf(lambda r_value: r_score(r_value), StringType())
f_udf = udf(lambda f_value: f_score(f_value), StringType())
m_udf = udf(lambda m_value: m_score(m_value), StringType())
```

```
from pyspark.sql.functions import concat

# Create RFM Segmentation DataFrame
rfm_seg = (
    rfm.withColumn('R Score', r_udf('Recency'))
    .withColumn('F Score', f_udf('Frequency'))
    .withColumn('M Score', m_udf('Monetary'))
    .withColumn('RFM Score', concat(col('R score'), col('F Score'), col('M Score')))
)

display(rfm_seg)
```

```
# Statistical Summary of RFM scores
rfm_seg.groupBy('RFM Score') \
    .agg({'Recency': 'mean', 'Frequency': 'mean', 'Monetary': 'mean'}) \
    .sort('RFM Score').show(10)
```
